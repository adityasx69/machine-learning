```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/data.csv')
```

```python
type_count = df["Type1"].value_counts(ascending=True)
```

```python
plt.barh(type_count.index,type_count,color="#03dffc",edgecolor="black")
plt.title("no of pokemon by type")
plt.xlabel("count")
plt.ylabel("type")
plt.tight_layou
```
