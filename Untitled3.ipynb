```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories = ["Freshman","Sophomores","Juniors","Seniors"]
```

```python
values = np.array([300,250,275,225])
```

```python
color = ["yellow","green","red","pink"]
```

```python
plt.pie(values,labels=categories,autopct="%1.1f%%",colors=color,explode=[0,0,0,0.1],shadow=True,startangle=0)
plt.title("College")
```
