```python
import numpy as np
```

```python
x = np.array([2023,2024,2025,2026])
```

```python
y1 = np.array([15,25,30,20])
```

```python
y2 = np.array([17,23,38,5])
```

```python
y3 = np.array([13,15,20,30])
```

```python
line_style = dict(marker=".",markersize=30,markerfacecolor="#1cd3fc",markeredgecolor="red",linestyle="solid",linewidth=4)
```

```python
plt.title("Class Size",fontsize=25,fontweight="bold",color="#2d4cfc")
plt.xlabel("Year",fontsize=20,fontweight="bold",color="#2dbefc")
plt.ylabel("Students",fontsize=20,fontweight="bold",color="#2dbefc")
plt.tick_params(axis="both",colors="#2dbefc")
plt.plot(x,y1,color="#1c5bfc",**line_style)
plt.plot(x,y2,color="green",**line_style)
plt.plot(x,y3,color="red",**line_style)
plt.xticks(x)
```
